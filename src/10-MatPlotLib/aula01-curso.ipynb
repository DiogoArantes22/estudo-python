# Matplotlib 

```python
import matplotlib.pyplot as plt
import numpy as np 

```

```python
t = np.linspace(0,2*np.pi,100)
y = np.cos(t)
print(f't: {t}')
print(f'y: {y}')
```

```python
plt.plot(t,y)
plt.show()
```

```python
t = np.linspace(0,2*np.pi,500)
y = np.cos(4*t)
plt.plot(t,y)
plt.title('Gráfico do Cosseno')
plt.xlabel('Tempo(s)')
plt.ylabel('Amplitude')
plt.show()
```

```python
plt.figure('Cosseno', figsize=(6,7)) #Figure : Define o nome da figura e o tamanho
plt.plot(t,y) #Plot : Plota o gráfico
plt.title('Gráfico do Cosseno') #Title : Define o título do gráfico
plt.xlabel('Tempo(s)') #Xlabel : Define o rótulo do eixo x
plt.ylabel('Amplitude') #Ylabel : Define o rótulo do eixo y
plt.show() #Show : Exibe o gráfico

y1 = np.sin(4*t)
plt.figure(2)
plt.plot(t,y1)
plt.title('Gráfico do Seno')
plt.xlabel('Tempo 2 (s)')
plt.ylabel('Amplitude 2')
plt.show()
```

```python
t1 = np.arange(0,100)
y2 = np.exp(t1)

plt.figure('Grafico', figsize=(10,5))
plt.plot(t1,y2)
plt.title('Gráfico do Exponencial')
plt.xlabel('Tempo 3 (s)')
plt.ylabel('Amplitude 3')
plt.show()
```

```python
t = np.linspace(0, 2*np.pi, 100)
y = np.cos(t)
y1 = np.sin(t)

plt.figure('Grafico', figsize=(10,5))
plt.plot(t,y)
plt.plot(t,y1)
plt.title('Gráfico do Cosseno e Seno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')
plt.grid()
plt.show()
```

## aula 04 - SubPlot

```python
x = np.linspace(0, 2*np.pi, 500)
c = np.cos(x)
s = np.sin(x)

plt.subplot(1,2,1) #subplot : Define o número de linhas, colunas e a posição do gráfico
plt.plot(x,c)
plt.title('Gráfico do Cosseno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.subplot(1,2,2)
plt.plot(x,s)
plt.title('Gráfico do Seno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.show()
```

```python
plt.subplot(2,1,1) #subplot : Define o número de linhas, colunas e a posição do gráfico
plt.plot(x,c)
plt.title('Gráfico do Cosseno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.subplot(2,1,2)
plt.plot(x,s)
plt.title('Gráfico do Seno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.show()
```

```python
plt.subplot(2,2,1) #subplot : Define o número de linhas, colunas e a posição do gráfico
plt.plot(x,c)
plt.title('Gráfico do Cosseno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.subplot(2,2,2)
plt.plot(x,s)
plt.title('Gráfico do Seno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.subplot(2,2,3) #subplot : Define o número de linhas, colunas e a posição do gráfico
plt.plot(x,c)
plt.title('Gráfico do Cosseno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')

plt.subplot(2,2,4)
plt.plot(x,s)
plt.title('Gráfico do Seno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Ampplitude')
plt.show()
```

```python
plt.figure('Grafico', figsize=(7,4))
plt.subplots_adjust(
    left=0.152,
    right=0.943,
    top=0.9,
    bottom=0.14,
    wspace=0.438,
    hspace=0.4) #Subplots_adjust : Define o espaçamento entre os gráficos

ax1 = plt.subplot(1,2,1) #subplot : Define o número de linhas, colunas e a posição do gráfico
plt.plot(x,c)
ax1.set_title('Gráfico do Cosseno')
ax1.set_xlabel('Eixo Tempo')
ax1.set_ylabel('Eixo Ampplitude')

ax2 = plt.subplot(1,2,2)
plt.plot(x,s)
ax2.set_title('Gráfico do Seno')
ax2.set_xlabel('Eixo Tempo')
ax2.set_ylabel('Eixo Ampplitude')


plt.show()
```

```python
x = np.arange(0,5,0.1)
y = x**2
y2= x**5

#SubPlots: Diferente do subplot, o subplots cria uma figura com vários gráficos, onde cada gráfico é um objeto do tipo Axes. O subplot cria uma figura com vários gráficos, onde cada gráfico é um objeto do tipo Axes e é possível acessar cada gráfico individualmente para personalizar suas características. O subplots é mais flexível e permite criar gráficos mais complexos, enquanto o subplot é mais simples e rápido de usar para gráficos simples.

fig, ((ax1, ax2),(ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(8,4))
ax1.plot(x,y)
ax2.plot(x,y2)
ax3.plot(x,y)
ax4.plot(x,y2)
plt.show()


```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(8,4))
axes[0,0].plot(x,y)
axes[0,1].plot(x,y2)
axes[1,0].plot(x,y)
axes[1,1].plot(x,y2)
plt.show()
```

```python
x = np.linspace(0, 2*np.pi, 70)
y = np.cos(4*x)

plt.figure('Grafico', figsize=(8,4))
plt.plot(x,y, color='r', lw=1.5, marker='*', linestyle='dotted') #Plot : Plota o gráfico, color : Define a cor do gráfico, lw : Define a largura da linha do gráfico
plt.grid(True)
plt.title('Gráfico do Cosseno')
plt.xlabel('Eixo Tempo')
plt.ylabel('Eixo Amplitude')
plt.show()
```

## Customizar eixos x, y

```python
x = np.linspace(0, 2*np.pi, 300)
y = np.cos(3*x)

fig, axe = plt.subplots( figsize=(7,4))
axe.plot(x,y)

axe.set_title('Gráfico do Cosseno', fontsize = 16)
axe.set_xlabel('Eixo Tempo',fontsize = 14)
axe.set_ylabel('Eixo Amplitude',fontsize = 14)

plt.xticks(np.arange(0, 2*np.pi+0.5, 0.5))
plt.yticks(np.arange(-1, 1.2, 0.2))

plt.grid(True)
plt.show()
```

## Linhas de Marcações 

```python
x = np.linspace(1,5,500)
y = np.log10(x)

plt.style.use("ggplot")

fig, axe = plt.subplots( figsize=(7,4))
axe.plot(x,y, lw= 1.2)
axe.plot([0,2.5],[0.4,0.4], color='gray', linestyle="--",lw =0.8)
axe.plot([2.5,2.5],[0,0.4], color='gray', linestyle="--",lw =0.8)
axe.plot(2.5,0.4, marker='o', color='black')
axe.set_xticks(np.arange(1,5.5,0.5))
plt.show()
```

## Texto no Gráfico 

```python
x = np.linspace(1,5,500)
y = np.log10(x)

plt.style.use("ggplot")

fig, axe = plt.subplots( figsize=(7,4))
axe.plot(x,y, lw= 1.2)

axe.text(2.6,0.35, 'P(2,5; 0,4)', color='black')
axe.text(3,0.42, "Logaritmo $y = log _{10}x$", fontsize = 10, color= 'black', bbox=dict(facecolor='red', alpha=0.5))
axe.annotate("P(2,5; 0,4)", xy=(2.5, 0.4),color='black', xytext=(0.5, 0.5), arrowprops=dict(facecolor='black', shrink=0.05))

axe.plot([0,2.5],[0.4,0.4], color='gray', linestyle="--",lw =0.8)
axe.plot([2.5,2.5],[0,0.4], color='gray', linestyle="--",lw =0.8)
axe.plot(2.5,0.4, marker='o', color='black')
axe.set_xticks(np.arange(0,5.5,0.5))
axe.set_title("Gráfico do Logaritmo")
axe.set_xlabel('Eixo Tempo',fontsize = 14)
axe.set_ylabel('Eixo Amplitude',fontsize = 14)
plt.show()
```
