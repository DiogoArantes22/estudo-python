```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
import matplotlib.style as st

df = pd.read_csv('../metrics.csv')
```

```python
epocas = range(1, len(df) + 1)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epocas, df["train_acc"], label="train")
plt.plot(epocas, df["val_acc"], label="valid")
plt.title("model accuracy")
plt.xlabel("epoch")
plt.ylabel("accuracy")
plt.grid(False)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(epocas, df["train_loss"], label="train")
plt.plot(epocas, df["val_loss"], label="valid")
plt.title("model loss")
plt.xlabel("epoch")
plt.ylabel("loss")
plt.grid(False)
plt.legend()

plt.tight_layout()
plt.show()
```
