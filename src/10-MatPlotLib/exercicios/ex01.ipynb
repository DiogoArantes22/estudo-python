```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
import matplotlib.style as st

df = pd.read_csv('../classification_results_trial_0001.csv')
```

```python
cont = df['predicted_class'].value_counts()
cont
```

```python
cont_real = df["real_class"].value_counts()
plt.figure(figsize=(6, 4))
plt.bar(cont_real.index, cont_real.values)
plt.title("Contagem por real_class")
plt.xlabel("Classe real")
plt.ylabel("Quantidade")
plt.show()
```

```python
cont_pred = df["predicted_class"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(cont_pred.index, cont_pred.values)
plt.title("Contagem por predicted_class")
plt.xlabel("Classe predita")
plt.ylabel("Quantidade")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df["prob_benign"], bins=20, edgecolor="black")
plt.title("Histograma de prob_benign")
plt.xlabel("Probabilidade de benign")
plt.ylabel("Frequência")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df["prob_malign"], bins=20, edgecolor="black")
plt.title("Histograma de prob_malign")
plt.xlabel("Probabilidade de malign")
plt.ylabel("Frequência")
plt.show()
```

## Scatter plot: X=prob_benign, Y=prob_malign, com cor diferente para acerto vs erro.

```python
acertos = df[df["acerto"] == True]
erros = df[df["acerto"] == False]

plt.figure(figsize=(7, 5))
plt.scatter(acertos["prob_benign"], acertos["prob_malign"], label="Acerto", alpha=0.7)
plt.scatter(erros["prob_benign"], erros["prob_malign"], label="Erro", alpha=0.7)
plt.title("prob_benign x prob_malign")
plt.xlabel("prob_benign")
plt.ylabel("prob_malign")
plt.legend()
plt.show()
```

## Qual erro é mais comum (FP ou FN)? Crie um gráfico que ajude a visualizar  isso. 

```python
erros_fp_fn = df[df["tipo_erro"].isin(["FP", "FN"])]["tipo_erro"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(erros_fp_fn.index, erros_fp_fn.values)
plt.title("Comparação entre falsos positivos (FP) e falsos negativos (FN)")
plt.xlabel("Tipo de erro")
plt.ylabel("Quantidade")
plt.tight_layout()
plt.show()
```

##  E em contexto médico: qual é mais preocupante (FN ou FP) e por quê?

O erro mais preocupante é o falso positivo, isso porque no falso negativo um caso realmente maligno é classificado como benigno,o que pode atrasar diagnóstico e tratamento, ou ser "tarde de mais"
